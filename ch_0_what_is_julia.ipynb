## 0-3 Julia의 중요한 설계 목표

Julia의 공식적인 설계 철학을 이해할 때 중요한 것이 있다.<br>

Julia는 단순히<br>

"Python보다 빠른 Python"

을 만들려고 한 언어가 아니다.<br>

핵심은 수학적/과학적 계산을 자연스럽게 표현하면서, 그 코드 자체를 컴파일러가 효율적으로 실행할 수 있게 하는 것이다.<br>

이를 위해 Julia에는 뒤에서 배울 중요한 개념들이 등장한다

**Multiple Dispatch -> Type System -> Specialization -> JIT Compilation -> Native Code**

지금 당장 이 개념들을 완전히 이해할 필요는 없다.

다만 앞으로 Julia를 배울 때 다음 관계를 기억해두면 좋다.

## 0-4 Julia는 "인터프리터 언어"인가?

Python을 사용해왔다면 상당히 중요한 질문이다.

보통 처음에는 이렇게 생각하기 쉽다.
*Python = interpreted*
*C = compiled*
그리고

"Julia는 어느 쪽인가?"

라는 질문이 생긴다.

Julia는 단순히 둘 중 하나라고 말하기 어렵다.

Julia 코드는 대략적으로
*Julia source code -> Julia compiler -> LLVM -> Machine code -> CPU*
라는 과정을 거친다.

특히 Julia는 **JIT(Just-In-Time) compilation**을 적극적으로 사용한다.

즉 프로그램을 실행하면서 필요한 함수의 구체적인 타입에 맞춰 코드를 컴파일한다.

예를 들어

```julia
function square(x)
    return x * x
end
```

라는 함수가 있다고 하자.

처음에는 square라는 함수의 코드만 존재한다.

그런데 *square(3)*
을 호출하면 **x**가 **Int**라는 사실을 이용해 적절한 코드를 생성할 수 있다.
그리고
*square(3.14)* 를 호출하면 Float64에 맞는 코드를 생성할 수 있다.

이것이 Julia의 **multiple dispatch**와 **specialization**을 이해하는 중요한 출발점이다.

## 0-8 Julia의 핵심 특징: Multiple Dispatch  
지금은 개념만 잡자  
예를 들면

```julia
f(x) = ...
```

이라는 함수가 있을때 julia에서는 같은 이름의 함수에 여러 method를 정의할 수 있다.

```julia
f(x+::Int) = x + 1
f(::Float64) = x + 2
f(x::String) = length(x)
```

그러면

```julia
f(10)
```

```julia
f(10.0)
```

```julia
f("Julia")
```

는 서로 다른 method를 사용한다<br>
이것이 **multiple dispatch**의 아주 단순한 예다.<br>
이것은 **타입 -> 함수 -> Multiple Dispatch -> Generic numerical programming**  
라는 설계와 연결되있어 나중에 수치해석 알고리즘을 구현할때 중요해진다


## 0-9 Python과 가장 큰 사고방식 차이 중 하나

Python에서는 흔히

```julia
class Something:
    def operation(self):
        ...
```

처럼 **객체가 자신의 동작을 가지고 있는 방식**으로 프로그램을 설계하는 경우가 많다  
Julia에서는 보다


```julia
operation(x)
```

이라는 함수 중심의 설계가 강하다.  
그리고 타입에 따라 적절한 method가 선택된다.

```julia
operation(x::A)
operation(x::B)
operation(x::C)
```

이 구조는 나중에 과학 계산에서 매우 유용하다.  

예를 들어 어떤 수치 알고리즘이 있다고 하자.

```julia
solve(problem)
```

이라는 인터페이스를 만들고,

```julia
solve(ODEProblem)
solve(OptimizationProblem)
solve(LinearSystem)
```

처럼 다양한 문제 유형에 대응시키는 것이 자연스럽다.
